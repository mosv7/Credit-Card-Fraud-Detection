```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler #Safer than MinMaxScaler and StandardScaler

X = df.drop('Class', axis=1)
y = df['Class']

X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)

X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=(0.15 / 0.85), random_state=42, stratify=y_temp)

scaler = RobustScaler()

X_train_scaled = X_train.copy()
X_val_scaled = X_val.copy()
X_test_scaled = X_test.copy()

X_train_scaled[['Time', 'Amount']] = scaler.fit_transform(X_train[['Time', 'Amount']])
X_val_scaled[['Time', 'Amount']] = scaler.transform(X_val[['Time', 'Amount']])
X_test_scaled[['Time', 'Amount']] = scaler.transform(X_test[['Time', 'Amount']])

print(f'Train shape: {X_train_scaled.shape}')
print(f'Validation shape: {X_val_scaled.shape}')
print(f'Test shape: {X_test_scaled.shape}')

print('\n--- Fraud Percentage ---')
print(f'Train Fraud Ratio: {y_train.mean() * 100:.4f}%')
print(f'Validation Fraud Ratio: {y_val.mean() * 100:.4f}%')
print(f'Test Fraud Ratio: {y_test.mean() * 100:.4f}%')
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, average_precision_score

baseline_lr = LogisticRegression(max_iter=1000, random_state=42)
baseline_lr.fit(X_train_scaled, y_train)

y_val_pred = baseline_lr.predict(X_val_scaled)
y_val_probs = baseline_lr.predict_proba(X_val_scaled)[:, 1]

print('=== Baseline Logistic Regression Results (Threshold = 0.5) ===')
print(confusion_matrix(y_val, y_val_pred))

print('\n Classification Report:')
print(classification_report(y_val, y_val_pred, digits=4))

pr_auc = average_precision_score(y_val, y_val_probs)
print(f'\n Precision-Recall AUC: {pr_auc:.4f}')
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_val, y_val_probs)

f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)  # Adding a small value to avoid division by zero

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]
best_f1 = f1_scores[best_idx]

print(f'Optimal Threshold on Validation Set: {best_threshold:.4f}')
print(f'Best F1 Score: {best_f1:.4f}')

y_val_pred_optimal = (y_val_probs >= best_threshold).astype(int)

print('\n=== Optimized Threshold Logistic Regression Results ===')
print('\nConfusion Matrix:')
print(confusion_matrix(y_val, y_val_pred_optimal))

print('\nClassification Report:')
print(classification_report(y_val, y_val_pred_optimal, digits=4))

plt.figure(figsize=(10, 5))
plt.plot(thresholds, precision[:-1], label='Precision', linestyle='--')
plt.plot(thresholds, recall[:-1], label='Recall', linestyle='--')
plt.plot(thresholds, f1_scores[:-1], label='F1 Score', color='red', linewidth=2)
plt.axvline(x=best_threshold, color='black', linestyle=':', label=f'Best Threshold = {best_threshold:.4f}')

plt.title('Precision, Recall and F1 Score vs Threshold', fontsize=14)
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train_scaled, y_train)

print(f'Original Train Class Distribution: {dict(y_train.value_counts())}')
print(f'SMOTE Train Class Distribution: {dict(y_train_smote.value_counts())}')

lr_smote = LogisticRegression(max_iter=1000, random_state=42)
lr_smote.fit(X_train_smote, y_train_smote)

y_val_probs_smote = lr_smote.predict_proba(X_val_scaled)[:, 1]

precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_probs_smote)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)

best_idx = np.argmax(f1_scores)
best_threshold_smote = thresholds[best_idx]
best_f1_smote = f1_scores[best_idx]

y_val_pred_smote = (y_val_probs_smote >= best_threshold_smote).astype(int)
pr_auc_smote = average_precision_score(y_val, y_val_probs_smote)

print(f'\n=== Logistic Regression + SMOTE Results ===')
print(f'Optimal Threshold: {best_threshold_smote:.4f}')
print(f'Best F1-Score:     {best_f1_smote:.4f}')
print(f'PR-AUC:            {pr_auc_smote:.4f}')

print('\nConfusion Matrix:')
print(confusion_matrix(y_val, y_val_pred_smote))

print('\nClassification Report:')
print(classification_report(y_val, y_val_pred_smote, digits=4))
```

```python
plt.figure(figsize=(10, 5))
plt.plot(thresholds, precisions[:-1], label='Precision', linestyle='--')
plt.plot(thresholds, recalls[:-1], label='Recall', linestyle='--')
plt.plot(thresholds, f1_scores[:-1], label='F1 Score', color='red', linewidth=2)
plt.axvline(x=best_threshold_smote, color='black', linestyle=':', label=f'Best Threshold = {best_threshold_smote:.4f}')

plt.title('Precision, Recall and F1 Score vs Threshold', fontsize=14)
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=50,
    max_depth=10,
    random_state=42,
    n_jobs=1
)

rf_model.fit(X_train_scaled, y_train)

y_val_probs_rf = rf_model.predict_proba(X_val_scaled)[:, 1]

pr_auc_rf = average_precision_score(y_val, y_val_probs_rf)

precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_probs_rf)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)

best_idx = np.argmax(f1_scores)
best_threshold_rf = thresholds[best_idx]
best_f1_rf = f1_scores[best_idx]

y_val_pred_rf = (y_val_probs_rf >= best_threshold_rf).astype(int)

print("=== Random Forest Results (n_estimators=50, max_depth=10) ===")
print(f"Optimal Threshold: {best_threshold_rf:.4f}")
print(f"Best F1-Score:     {best_f1_rf:.4f}")
print(f"PR-AUC:            {pr_auc_rf:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_val_pred_rf))

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred_rf, digits=4))

plt.figure(figsize=(10, 5))
plt.plot(thresholds, precisions[:-1], label='Precision', linestyle='--')
plt.plot(thresholds, recalls[:-1], label='Recall', linestyle='--')
plt.plot(thresholds, f1_scores[:-1], label='F1 Score', color='red', linewidth=2)
plt.axvline(x=best_threshold_rf, color='black', linestyle=':', label=f'Best Threshold = {best_threshold_rf:.4f}')

plt.title('Precision, Recall and F1 Score vs Threshold', fontsize=14)
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Grid search for hyperparameter tuning of Random Forest

max_depths = [5, 8, 10]
n_estimators = [10, 50, 100]
class_weights = [None, 'balanced', 'balanced_subsample']

best_rf_f1 = 0
best_rf_thresh = 0.5
best_rf_model = None
best_rf_params = {}

print('=== Starting Random Forest Grid Search ===')

for depth in max_depths:
    for n_est in n_estimators:
        for class_weight in class_weights:
            rf = RandomForestClassifier(
                n_estimators=n_est,
                max_depth=depth,
                class_weight=class_weight,
                random_state=42,
                n_jobs=1
            )
            rf.fit(X_train_scaled, y_train)

            probs = rf.predict_proba(X_val_scaled)[:, 1]

            precisions, recalls, thresholds = precision_recall_curve(y_val, probs)
            f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)

            best_idx = np.argmax(f1_scores)
            opt_thresh = thresholds[best_idx]
            max_f1 = f1_scores[best_idx]

            if max_f1 > best_rf_f1:
                best_rf_f1 = max_f1
                best_rf_thresh = opt_thresh
                best_rf_model = rf
                best_rf_params = {
                    'n_estimators': n_est,
                    'max_depth': depth,
                    'class_weight': class_weight
                }

print(f'Best Parameters Found: {best_rf_params}')
print(f'Optimal Threshold:     {best_rf_thresh:.4f}')
print(f'Bset Validation F1:    {best_rf_f1:.4f}')

best_rf_probs = best_rf_model.predict_proba(X_val_scaled)[:, 1]
best_rf_preds = (best_rf_probs >= best_rf_thresh).astype(int)
best_rf_pr_auc = average_precision_score(y_val, best_rf_probs)

print(f'Best RF PR-AUC:     {best_rf_pr_auc:.4f}')

print("\nBest Random Forest Confusion Matrix:")
print(confusion_matrix(y_val, best_rf_preds))

print("\nBest Random Forest Classification Report:")
print(classification_report(y_val, best_rf_preds, digits=4))
```

```python
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_recall_curve, classification_report, confusion_matrix, average_precision_score
import numpy as np

clf_lr = LogisticRegression(max_iter=1000, random_state=42)
clf_rf = RandomForestClassifier(n_estimators=50, max_depth=10, random_state=42, n_jobs=1)

voting_clf = VotingClassifier(
    estimators=[('lr', clf_lr), ('rf', clf_rf)],
    voting='soft'
)

voting_clf.fit(X_train_scaled, y_train)

y_val_probs_vc = voting_clf.predict_proba(X_val_scaled)[:, 1]

pr_auc_vc = average_precision_score(y_val, y_val_probs_vc)

precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_probs_vc)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)

best_idx = np.argmax(f1_scores)
best_thresh_vc = thresholds[best_idx]
best_f1_vc = f1_scores[best_idx]

y_val_pred_vc = (y_val_probs_vc >= best_thresh_vc).astype(int)

print('=== Voting Classifier Results (Soft Voting) ===')
print(f"Optimal Threshold: {best_thresh_vc:.4f}")
print(f"Best F1-Score:     {best_f1_vc:.4f}")
print(f"PR-AUC:            {pr_auc_vc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_val_pred_vc))

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred_vc, digits=4))
```

```python
from xgboost import XGBClassifier

clf_xgb = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_weight=3,
    gamma=0,
    reg_alpha=0.0,
    reg_lambda=1.0,
    scale_pos_weight=1,
    objective='binary:logistic',
    eval_metric='aucpr',
    random_state=42,
    n_jobs=-1
)

clf_xgb.fit(X_train_scaled, y_train)

y_val_probs_xgb = clf_xgb.predict_proba(X_val_scaled)[:, 1]

pr_auc_vc = average_precision_score(y_val, y_val_probs_xgb)

precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_probs_xgb)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)

best_idx = np.argmax(f1_scores)
best_thresh_vc = thresholds[best_idx]
best_f1_vc = f1_scores[best_idx]

y_val_pred_xgb = (y_val_probs_xgb >= best_thresh_vc).astype(int)

print('=== XGBoost Classifier Results ===')
print(f"Optimal Threshold: {best_thresh_vc:.4f}")
print(f"Best F1-Score:     {best_f1_vc:.4f}")
print(f"PR-AUC:            {pr_auc_vc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_val_pred_xgb))

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred_xgb, digits=4))
```
